Q. WAF to print the length of a list. (list is the parameter)

```python
cities = ["janakpur","kathmandu","chitwan","birjung"]
animal = ["tiger","lion","hourse","panda"]

def print_len(list):
    return len(list)

print(print_len(cities))
print(print_len(animal))
```

Q. WAF to print the elements of a list in a single line. (list is the parameter)


*   example:

*   print(animal[0])
*   print(animal[1])


*   like this but in function


```python
cities = ["janakpur","kathmandu","chitwan","birjung"]
animal = ["tiger","lion","hourse","panda"]

def print_list(list):
    for item in list:
        print(item , end=" ")

print_list(animal)
print_list(cities)
```

Q. WAF to find the factorialnof n. (n is the paramenter)


*   example:
*   4! = 1*2*3*4. like that for n factorial


```python
def fac(n):
    factorial = 1
    for i in range(1,n+1):
     factorial *= i
    return factorial

fac(5)
```

**RECURSION**

---



---



when a function calls itself repeatedly. basically we can say it's a advance version of Loops

```python
def show(n):
    if (n==0):
        return
    print(n)
    show(n-1)

show(5)
```
